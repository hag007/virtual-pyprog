# טווחים (`range`)

### אג'נדה
- הטיפוס והפונקציה טווח (`range`)
- אינדקסים בטווחים


הפונקציה `range` מאפשרת ליצור סדרה ממוינת של מספרים שלמים בתחום מסוים. התוצאה היא אובייקט שמייצר את המספרים לפי דרישה, כלומר המספרים נוצרים רק כאשר אנחנו צריכים אותם, ולא נשמרים בזיכרון כקבוצה שלמה.

זה מאוד שימושי כשמייצרים סדרות מספרים או כאשר רוצים לעבור על רשימות לפי אינדקסים.

#### תחביר
- `range(start, stop, step)` - מייצר מספרים מstart (כולל) עד stop (לא כולל) בקפיצות של step.
- `range(start, stop)`  - מייצר מספרים מstart (כולל) עד stop (לא כולל) בקפיצות של 1. כלומר, כשאין step, מניחים שמדובר בקפיצות של 1.
- `range(stop)`   מייצר מספרים מאפס (כולל) עד stop (לא כולל) בקפיצות של 1. כלומר, כשאין step וגם start, מניחים שמתחילים מ0 והולכים בקפיצות של 1.

`range` הוא טיפוס בפני עצמו.

```python
range(10)
```

כדי לראות את הערכים שrange מייצר, נצטרך להמיר אותו לרשימה.

```python
list(range(10))
```

```python
list(range(0,10,2))
```

`step` יכול להיות גם שלילי - אבל אז נצטרך שיתקיים `start>end`

```python
list(range(10,0,-2))
```

מה יקרה אם נבקש מפייתון טווח ריק?

```python
list(range(0, 10, -1))
```

כפי שראינו בחיתוך, סדרה ריקה תהפוך לרשימה ריקה.  
עם זאת, למה הטווח הבא לא עובד?

```python
range(0,10,0)
```

במקרה הזה לא מדובר בסדרה ריקה, אלא בסדרה אינסופית! אף פעם לא נגיע ל10, ולנצח נמשיך להחזיר 0.  
לכן מי שכתב את פייתון החליט -`step=0` יחזיר שגיאה.
