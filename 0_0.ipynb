בסרטון למדנו על משתנים - מהם, למה צריך אותם, ואיך משתמשים בהם. 

לכל משתנה יש:  
- **ערך** – התוכן, כלומר הנתונים עצמם. (לדוגמה: 5.2) 
- **שם** – כינוי המשמש אותנו כדי לגשת לערך (לדוגמה my_number).  
- **טיפוס (type)** – סוג המשתנה, שקובע איזה סוג של מידע הוא יכול להכיל (לדוגמה float).

לדוגמא:


```python
s = "bob"
r = True
age = 35

```



ב-Python יוצרים משתנים ע"י השמת ערך לשם, באמצעות סימן ה`=`.
שימו לב! זה אינו סימן ה= של מתמטיקה. משמעות סימן ה`=` היא "תשמור את הערך מימין, בשם משמאל".

טיפוסי המשתנים נקבעים אוטומטית לפי ערכים שהוכנסו אליהם והפעולות שבוצעו עליהם. כלומר, אם כתבתי ```a=5```, פייתון יודע ש5 הוא מספר, ובפרט - הוא יודע ש5 הוא `int`.

### משתנים והשמות – כללים ופעולות נוספות

#### שמות משתנים

פייתון לא מאפשר לכל רצף אותיות להיות שם משתנה.
שם משתנה חוקי בפייתון צריך לקיים את התנאים הבאים:
- מכיל רק אותיות באנגלית, מספרים ואת הסימן `_`.
- התו הראשון הוא לא מספר.

בנוסף, הכתיב המקובל עבור פייתון לשמות משתנים הוא באותיות קטנות, עם `_` בין מילים. לדוגמה: `my_variable`.

#### השמה למשתנה קיים ושינוי טיפוסי משתנים 
ניתן לשנות ערך של משתנה על ידי השמה מחדש. יתרה מזאת, ניתן אף להחליף את **סוג הערך (הטיפוס)** שנמצא במשתנה.
- על מנת לבדוק מהו הטיפוס של הערך הנמצא במשתנה, נשתמש בפקודה `type`, המחזירה את מחזירה את סוג המשתנה (או הביטוי)

````python
n=10
print(n, type(n)) 
n=11
print(n, type(n)) 
n = 1.3141
print(n, type(n)) 
````
