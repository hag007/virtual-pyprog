לאחר שהכרנו את הספרייה NumPy, שלימדה אותנו לעבוד עם נתונים על-ידי מערכים וחישובים וקטוריים, הגיע הזמן להכיר ספרייה נוספת וחזקה במיוחד – pandas.

## מהי ספרית pandas?

pandas היא ספרייה בפייתון שעוזרת לנו לעבוד עם **נתונים טבלאיים** בצורה נוחה.   
אפשר לחשוב עליה כמו אקסל בתוך פייתון – היא מאפשרת לשמור, לקרוא ולערוך נתונים שמסודרים בטבלה של שורות ועמודות.   
למשל, אם יש לנו קובץ עם ציונים של סטודנטים או נתונים על מחירים של מוצרים, אפשר בעזרת pandas לנתח אותם בקלות.

ברמה הטכנית, pandas מאפשרת לעשות הרבה דברים חשובים במעט מאוד שורות קוד: למיין נתונים, לסנן את המידע שמעניין אותנו, לחשב ממוצעים, לשלב בין כמה טבלאות ולנקות נתונים בעייתיים. זהו למעשה כלי בסיסי לאנליזות הנמצא בשימוש רחב במדעי נתונים.

## המחלקה `DataFrame`

ב-pandas המבנה המרכזי לעבודה עם נתונים נקרא `DataFrame`. מדובר במחלקה המייצגת טבלה דו־ממדית שמורכבת משורות ועמודות, בדומה לגיליון אקסל.  
כל עמודה או שורה בודדת מיוצגת באמצעות אובייקט מטיפוס `Series` - מחלקה המייצגת סדרה חד-מימדית של נתונים.
בהרבה מובנים, ניתן לחשוב על `Series` כמערך חד-מימדי, ועל`DataFrame` כמערך דו-מימדי, כאשר המימד/הציר הראשון (`axis=0`) הוא השורות והציר השני (`axis=1`) הוא העמודות.   

בניגוד למבני נתונים אחרים המתייחסים לטבלה כמטריצה אחידה, ב־pandas כל עמודה וכל שורה נחשבות ליחידות מידע עצמאיות, שלכל אחת מהן יכול להיות **טיפוס נתונים שונה ומשמעות שונה**.
לכן, ברוב המקרים לא נבצע פעולות על כל הטבלה כמכלול, אלא לאורך אחד מהצירים – עמודות או שורות – למשל, חישוב ממוצעים לכל עמודה, או סינון שורות שעומדות בתנאי מסוים.

### היתרונות של `DataFrame` לייצוג טבלה

כבר למדנו כיצד לייצג טבלאות באמצעות מבני נתונים בפייתון. מדוע צריך כעת מבנה נתונים חדש?

נחשוב רגע על ההבדל בין מטריצה (כמו מערך NumPy) לבין טבלה:
במטריצה, כל הערכים חייבים להיות מאותו **טיפוס נתונים יחיד** — לרוב מספרים.
לעומת זאת, בטבלה אמיתית (כמו גיליון נתונים), **כל עמודה יכולה להכיל טיפוס שונה**: עמודה אחת מספרית (למשל, "ציון"), אחרת טקסטואלית ("שם"), ואולי גם עמודה בוליאנית ("עבר/נכשל").

בנוסף, במטריצה אין משמעות מובנית לשורות או לעמודות — ההתייחסות לערכים נעשית רק לפי המיקום המספרי שלהם.
לעומת זאת, בטבלה יש **שמות ייחודיים לשורות ולעמודות**, המעניקים לנתונים משמעות סמנטית מעבר לסדרם.
שמות העמודות נשמרים בשדה `columns`, ושמות השורות — או האינדקס — נשמרים בשדה `index`.

```{admonition} **שימו לב!**  
:class: error
עד עכשיו הכרנו את המילה אינדקס כ**מיקום**.  
עם זאת, מקובל גם לקרוא ל**שמות** השורות אינדקס, גם אם אינן מספריות או אינן תואמות למיקום.

כן, זה באמת מבלבל. לצערנו, זה כל כך נפוץ שחייבים להשתמש במונחים האלו.


```

ניתן לחשוב על טבלה של pandas כעל שילוב בין רשימה דו־ממדית למילון:
מצד אחד, יש לה מבנה של שורות ועמודות שניתן לגשת אליהן לפי מיקום.
מצד שני, לכל עמודה ולכל שורה יש שם (כמו מפתח במילון) המאפשר גישה נוחה וברורה לפי משמעות הנתון ולא רק לפי מיקומו.


לסיכום:

| מאפיין            | מטריצה                           | טבלה (DataFrame / טבלה כללית)              |
|--------------------|----------------------------------|---------------------------------------------|
| **שמות לעמודות/שורות** | אין שמות, רק מיקום מספרי       | יש כותרות לעמודות ואינדקסים לשורות       |
| **סוגי נתונים**        | אחיד – כל הערכים מאותו סוג נתונים | מגוון – לכל עמודה יכול להיות סוג נתונים שונה |
| **שימוש עיקרי**        | חישובים מתמטיים ואלגברה ליניארית | ניתוח וארגון נתונים מהעולם האמיתי           |
| **גמישות**             | מוגבלת – כל המטריצה מכילה טיפוס בודד        | גבוהה – אפשר לשלב מספרים, טקסט, תאריכים וכו’|
| **גישה לערכים**        | לפי מיקום (שורה, עמודה)          | לפי מיקום **או** לפי שם עמודה/שורה          |
