## `break` ו `continue`

### מילת המפתח `break`
לפעמים אנחנו רוצים להפסיק לולאה עוד לפני שהיא סיימה את כל האיטרציות שלה. בדיוק בשביל זה קיימת מילת המפתח `break`.  
כשמופיעה הפקודה `break` בתוך לולאה, היא גורמת ליציאה מיידית מהלולאה הקרובה ביותר, בלי להמשיך להריץ את שאר הפקודות שלה ובלי להמשיך לאיטרציות נוספות.

```python
lst = [4, 2, -6, 3,-9]
for elem in lst:
    if elem < 0: 
        print("First negative number is", elem)
        break
    print("current number is: ", elem)
print("done")
```

אפשר לראות שלא הדפסנו את השורה `"current number is:..."` עבור -6, ולא עברנו על המשך הרשימה. במקום זאת, הקוד סיים את הלולאה ולכן עבר לפקודה `print("done")`.

**דוגמה:** שימוש ב־ `break` בלולאת `for` למציאת המחלק הראשוני הקטן ביותר. שימו לב להזחות!

```python
n=25
for div in range(2,int(n**0.5)+1): # Note the +1!
    if n % div == 0:
        break
if n % div == 0: 
    print(n, "is not prime")
else:
    print(n, "is prime")
```

### מילת המפתח `continue`
לפעמים לא נרצה לעצור לגמרי את הלולאה, אלא פשוט לדלג על האיטרציה הנוכחית ולהמשיך מיד לזו הבאה. בדיוק בשביל זה יש את מילת המפתח `continue`.

כאשר פוגשים ב־`continue`, פייתון מפסיק להריץ את שאר הפקודות של אותה איטרציה, וקופץ ישר להתחלת האיטרציה הבאה.

```python
lst = [4, 2, -6, 3,-9]
for elem in lst:
    if elem < 0: 
        print("First negative number is", elem)
        continue
    print("current number is: ", elem)
print("done")
```

אפשר לראות שלא הדפסנו את השורה `"current number is:..."` עבור -6 או -9, אבל כן המשכנו לעבור על הרשימה.

**דוגמה:** פונקציה שמקבלת רשימה, ומחזירה רשימה חדשה שבה הערכים ייחודיים (בלי חזרות).

```python
def unique_list(lst):
    uniques = []
    for x in lst: 
        if x in uniques: # If already in list, don't add again
            continue 
        uniques.append(x)
    return uniques

unique_list([1,4,5,8,3,5,7,1,2])
```

כתבו את הפונקציה has_negative שמקבלת רשימה של מספרים ומחזירה True אם יש ברשימה מספר שלילי, ו־False אחרת.
